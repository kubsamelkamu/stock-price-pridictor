```python
import pandas as pd
import os

raw_data_dir = '../data/processed'
raw_files = os.listdir(raw_data_dir)

def check_missing_values(file_path):
    """
        Read CSV file and check for missing values in each column.
    """
    if not os.path.exists(file_path):
        print('File not found:', file_path)
        return
    
    # Load the data set
    df = pd.read_csv(file_path)
    missing_values = df.isnull().sum()
    
    if missing_values.sum() > 0:
        print('Missing values in', file_path)
        print(missing_values)
    else:
        print('No missing values in', file_path)

for file in raw_files:
    file_path = os.path.join(raw_data_dir, file)
    check_missing_values(file_path) 
```

```python
import pandas as pd
import os


input_dir = "../data/processed"
output_dir = "../data/processed_cleaned"


os.makedirs(output_dir, exist_ok=True)


files = [
    "featured_engineered_AAPL.csv",
    "featured_engineered_AMZN.csv",
    "featured_engineered_GOOGL.csv",
    "featured_engineered_MSFT.csv",
    "featured_engineered_TSLA.csv"
]

def remove_missing_values(file_path, save_path):
    """
    Reads a CSV file, removes rows with missing values, and saves the cleaned dataset.
    """
    
    data = pd.read_csv(file_path)

    
    data_cleaned = data.dropna()

    # Save the cleaned dataset
    data_cleaned.to_csv(save_path, index=False)
    print(f"Removed missing values and saved to {save_path}")

# Process each file
for file in files:
    input_path = os.path.join(input_dir, file)
    output_path = os.path.join(output_dir, file)
    remove_missing_values(input_path, output_path)

print("All missing values removed and cleaned files saved successfully.")
```
